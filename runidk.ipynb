```python
import csv
import numpy as np

with open("data/dataset/x_train.csv", "r") as f: #ouvrir un fichier csv, le lire et récupérer l'entête
    header = next(csv.reader(f)) #recupere seule l'entête du fichier csv
names = header[1:]   # on enlève "Id" car on ne veut pas l'analyser
assert len(names) == x_train.shape[1]

rows = []
for j, name in enumerate(names):
    col = x_train[:, j]
    valid = col[~np.isnan(col)]
    n_unique = len(np.unique(valid))
    rows.append([
        j, name,
        round(np.isnan(col).mean(), 3),          # la part de cases vides. Python compte True comme 1 et False comme 0, 
        #donc la moyenne donne directement le pourcentage de vides. round(..., 3) arrondit à 3 chiffres après la virgule.
        n_unique,                                 # nombre de valeurs distinctes
        valid.min() if len(valid) else "",
        valid.max() if len(valid) else "",
        round(valid.mean(), 3) if len(valid) else "",#si la colonne a au moins une réponse, calcule ; sinon, laisse la case vide
    ])

with open("resume_colonnes.csv", "w", newline="") as f: #On crée un nouveau fichier resume_colonnes.csv. Le "w" veut dire qu’on l’ouvre pour écrire dedans.
    w = csv.writer(f)
    w.writerow(["index", "nom", "nan_ratio", "n_unique", "min", "max", "moyenne"])
    w.writerows(rows)
```

```python
def resume(nom):
    j = names.index(nom)            # adapte si names est un array numpy
    col = x_train[:, j]
    print(f"--- {nom} ---")
    print(f"valeurs manquantes : {np.isnan(col).mean():.1%}")
    valeurs, comptes = np.unique(col[~np.isnan(col)], return_counts=True)
    if len(valeurs) <= 15:          # peu de valeurs : probablement des catégories
        for v, c in zip(valeurs, comptes):
            print(f"  {v:>6} : {c} personnes")
    else:                           # beaucoup de valeurs : probablement numérique
        print(f"  {len(valeurs)} valeurs différentes, de {valeurs.min()} à {valeurs.max()}")

resume("GENHLTH")
resume("CHILDREN")
```

```python
code_families = {
    9:    [7, 8, 9],
    99:   [77, 88, 99],
    999:  [777, 888, 999],
    9999: [7777, 8888, 9999],
}

rows = []
for j, name in enumerate(names):
    col = x_train[:, j]
    values = col[~np.isnan(col)]
    if len(values) == 0:
        rows.append((name, 1.0, 0, np.nan, np.nan, np.nan, []))
        continue
    uniques = np.unique(values)
    vmax = uniques.max()

    codes = []
    real_max = vmax
    if vmax in code_families:
        family = code_families[vmax]
        real_values = uniques[~np.isin(uniques, family)]
        # it's only a code if all real values are below the smallest code
        if len(real_values) > 0 and real_values.max() < min(family):
            codes = [c for c in family if c in uniques]
            real_max = real_values.max()

    rows.append((name, np.isnan(col).mean(), len(uniques),
                 uniques.min(), vmax, real_max, codes))

print(f"{'column':<12} {'%NaN':>6} {'n_vals':>7} {'min':>8} {'max':>10} {'real_max':>9}  codes")
for name, nan, n_vals, vmin, vmax, real_max, codes in rows:
    print(f"{name:<12} {nan:>6.1%} {n_vals:>7} {vmin:>8} {vmax:>10} {real_max:>9}  {codes}")
```

```python
from collections import defaultdict

groups = defaultdict(list)
for name, nan, n_vals, vmin, vmax, real_max, codes in rows:
    if nan > 0.9:
        groups["mostly missing (>90% NaN)"].append(name)
    elif n_vals <= 1:
        groups["constant"].append(name)
    elif real_max <= 2 and len(codes) > 0:
        groups["binary with codes (1/2 + 7/9)"].append(name)
    elif real_max <= 2:
        groups["binary without codes"].append(name)
    elif real_max <= 10 and len(codes) > 0:
        groups["small scale with codes"].append(name)
    elif len(codes) > 0:
        groups["numeric with codes"].append(name)
    else:
        groups["no codes detected"].append(name)

for group, cols in groups.items():
    print(f"\n=== {group} ({len(cols)} columns) ===")
    print(", ".join(cols))
```

```python
def clean(x, names, groups):
    x = x.copy()
    idx = {n: i for i, n in enumerate(names)}

    def set_nan(col, values):
        if col in idx:
            j = idx[col]
            x[np.isin(x[:, j], values), j] = np.nan

    def replace(col, old, new):
        if col in idx:
            j = idx[col]
            x[x[:, j] == old, j] = new

    # --- binary with codes: 7/9 -> NaN, 2 -> 0
    for c in groups["binary with codes (1/2 + 7/9)"]:
        set_nan(c, [7, 9])
        replace(c, 2, 0)

    # --- binary without codes: 2 -> 0
    for c in groups["binary without codes"]:
        replace(c, 2, 0)

    # --- counts of days / drinks / children: 88 -> 0, 77/99 -> NaN
    for c in ["PHYSHLTH", "MENTHLTH", "POORHLTH", "CHILDREN",
              "AVEDRNK2", "DRNK3GE5", "MAXDRNKS"]:
        replace(c, 88, 0)
        set_nan(c, [77, 99])

    # --- small scales: 7/9 -> NaN (with exceptions handled below)
    special = {"CHECKUP1", "DIABETE3", "BPHIGH4", "INCOME2"}
    for c in groups["small scale with codes"]:
        if c not in special:
            set_nan(c, [7, 9])

    set_nan("INCOME2", [77, 99])

    set_nan("CHECKUP1", [7, 9])
    replace("CHECKUP1", 8, 5)          # 8 = never -> after "5+ years"

    # DIABETE3 / BPHIGH4: 1 = yes, 2 = only during pregnancy, 3 = no, 4 = borderline
    for c in ["DIABETE3", "BPHIGH4"]:
        if c in idx:
            j = idx[c]
            old = x[:, j].copy()
            new = np.full_like(old, np.nan)
            new[old == 1] = 1
            new[old == 4] = 0.5
            new[np.isin(old, [2, 3])] = 0
            x[:, j] = new

    # --- hidden codes in "no codes detected"
    set_nan("DIABAGE2", [98, 99])
    set_nan("_AGEG5YR", [14])
    set_nan("_AGE65YR", [3])
    set_nan("DROCDY3_", [900])
    set_nan("_DRNKWEK", [99900])
    set_nan("MAXVO2_", [99900])

    # --- columns to drop
    drop = set(groups["mostly missing (>90% NaN)"]) | set(groups["constant"])
    drop |= {
        # admin / phone
        "_STATE", "FMONTH", "IDATE", "IMONTH", "IDAY", "IYEAR", "DISPCODE",
        "SEQNO", "_PSU", "QSTVER", "QSTLANG", "MSCODE", "NUMADULT", "NUMMEN",
        "NUMWOMEN", "PVTRESD1", "CELLFON3", "CADULT", "PVTRESD2", "CSTATE",
        "LANDLINE", "NUMHHOL2", "HHADULT",
        # survey weights
        "_STSTR", "_STRWT", "_RAWRAKE", "_WT2RAKE", "_CLLCPWT", "_LLCPWT",
        # complex formats with a clean calculated version
        "WEIGHT2", "HEIGHT3", "FRUITJU1", "FRUIT1", "FVBEANS", "FVGREEN",
        "FVORANG", "VEGETAB1", "ALCDAY5", "EXEROFT1", "EXEROFT2", "STRENGTH",
        "EXERHMM1", "EXERHMM2",
        # activity types, places, dates
        "EXRACT11", "EXRACT21", "IMFVPLAC", "FLSHTMY2", "HIVTSTD3",
    }
    keep = [j for j, n in enumerate(names) if n not in drop]
    kept_names = [names[j] for j in keep]
    return x[:, keep], kept_names


x_train_clean, kept_names = clean(x_train, names, groups)
x_test_clean, _ = clean(x_test, names, groups)
print("Columns kept:", len(kept_names), "out of", len(names))
```

```python
# 2. Filter again (cleaning created new NaNs)
nan_ratio = np.isnan(x_train_clean).mean(axis=0)
n_unique = np.array([len(np.unique(c[~np.isnan(c)])) for c in x_train_clean.T])

keep = (nan_ratio < 0.9) & (n_unique > 1)
x_train_clean = x_train_clean[:, keep]
x_test_clean = x_test_clean[:, keep]          # same columns for test!
kept_names = [n for n, k in zip(kept_names, keep) if k]
print("Columns kept:", len(kept_names))

# 3. Labels as 0/1
y01 = np.where(y_train == 1, 1, 0)   # converts -1/1 labels into 0/1

# 4. Correlation ranking (exploration only)
scores = []
for j in range(x_train_clean.shape[1]):
    col = x_train_clean[:, j].copy()
    col[np.isnan(col)] = np.nanmean(col)    # fill NaNs with the column mean
    if col.std() > 0:                       # skip constant columns (division by zero)
        scores.append((abs(np.corrcoef(col, y01)[0, 1]), kept_names[j]))

for s, name in sorted(scores, reverse=True)[:20]:
    print(f"{name:<12} {s:.3f}")
```

```python
for s, nom in sorted(scores)[:20]:   # les 20 scores les plus faibles, questions les moins corrélées avec la réponse 1, donc avec l'état malade
    print(f"{nom:<12} {s:.3f}")
```
